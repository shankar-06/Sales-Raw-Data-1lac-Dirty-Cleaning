### Dropping Duplicates

```python
df = df.drop_duplicates()
```

```python
df.shape
```

### Standardize the messy text columns

```python
df['Region'] = df['Region'].str.strip().str.title()
```

```python
df['PaymentMethod'] = df['PaymentMethod'].str.strip().str.title()
```

```python
payment_mapping = {'Paypal':'PayPal','Cod':'Cash on Delivery','Upi':'UPI'}
```

```python
df['PaymentMethod'] = df['PaymentMethod'].replace(payment_mapping)
```

### Standardize the numerical columns

```python
valc()
```

```python
df['Quantity'] = df['Quantity'].str.strip().str.replace('pcs','')
```

```python
df['Quantity'] = pd.to_numeric(df['Quantity'], errors = 'coerce').round().astype('Int64')
```

```python
df['UnitPrice'] = df['UnitPrice'].str.replace('$','')
```

```python
df['UnitPrice'] = pd.to_numeric(df['UnitPrice'], errors = 'coerce')
```

### Standardize the datetime columns

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'], errors = 'coerce', format = 'mixed')
```

### Finding Values for Null Values in numeric columns

```python
valc()
```

```python
df['UnitPrice'] = df['UnitPrice'].fillna(df.groupby('Product')['UnitPrice'].transform('first'))
```

```python
df['DiscountPercent'] = df['DiscountPercent'].fillna(0.0)
```

```python
mask_total = df['TotalSales'].isnull() & df['UnitPrice'].notnull() & df['Quantity'].notnull()
```

```python
discount_multiplier = (1 - df.loc[mask_total, 'DiscountPercent'] / 100)
```

```python
df.loc[mask_total, 'TotalSales'] = df.loc[mask_total, 'UnitPrice'] * df.loc[mask_total, 'Quantity'] * discount_multiplier
```

```python
mask_quantity = df['Quantity'].isnull() & df['UnitPrice'].notnull() & df['TotalSales'].notnull()
```

```python
df.loc[mask_quantity, 'Quantity'] = (df.loc[mask_quantity, 'TotalSales'] / df.loc[mask_quantity, 'UnitPrice']).round().astype('Int64')
```

```python
df = df[(df['Quantity'] > 0) | (df['Quantity'].isnull())]
```

### Memory Optimization

```python
df['Category'] = df['Category'].astype('category')
```

```python
df['Region'] = df['Region'].astype('category')
```

```python
df['SalesChannel'] = df['SalesChannel'].astype('category')
```

```python
df['PaymentMethod'] = df['PaymentMethod'].astype('category')
```

```python
df.info()
```

```python
valc()
```

```python
df['PaymentMethod'] = df['PaymentMethod'].replace({'Cash On Delivery': 'Cash on Delivery'})
```

### Clean up empty categories (like the capitalized 'Cash On Delivery')

```python
df['PaymentMethod'] = df['PaymentMethod'].cat.remove_unused_categories()
```

### Safely Handle Final Categorical Nulls

```python
rcol = ['Region','Country','PaymentMethod','SalesRep','CustomerName']
for col in rcol:
    if df[col].dtype.name == 'category':
        df[col] = df[col].cat.add_categories('Unknown')
    df[col] = df[col].fillna('Unknown')
```

### Drop the final stubborn rows missing math/dates

```python
df = df.dropna(subset = ['Quantity','OrderDate'])
```

### Export the Cleaned Dataset

```python
df.to_csv('Cleaned_1lac_Sales_Data.csv', index = False)
print('Export Done')
```
